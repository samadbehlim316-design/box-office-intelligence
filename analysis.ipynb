````python
# Movie Data Cleaning Process

## 1. Import Required Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import requests
```

## 2. Load the Excel Dataset

```python
df = pd.read_excel("top_rated_movies_final.xlsx")

print(df)
```

The dataset initially contains:

* **18,520 rows**
* **13 columns**

---

## 3. Check Dataset Columns

```python
print(df.columns)
```

### Columns

```text
ID
Title
Overview
Release Date
Popularity
Vote Average
Vote Count
Category
Production Budget (USD M)*
Box Office Collection (USD M)*
Oscar Nominated
Film Industry
Director
```

---

## 4. Check Duplicate Movie IDs

```python
print(df["ID"].value_counts())
```

The dataset contains duplicate movie IDs.

For example:

* ID `389015` appears 4 times.
* IDs such as `278`, `238`, `240`, and `424` appear twice.

This indicates that some movies are duplicated in the dataset.

---

## 5. Remove Duplicate IDs

Since `ID` is the unique identifier for a movie, duplicate IDs should be removed.

```python
df = df.drop_duplicates(subset="ID")

print(df)
```

After removing duplicate IDs:

```text
9959 rows × 13 columns
```

---

## 6. Verify Duplicate IDs

```python
print(df["ID"].value_counts())
```

After cleaning, every movie ID appears only once.

This confirms that duplicate movie IDs have been successfully removed.

---

## 7. Check Duplicate Movie Titles

```python
print(df["Title"].value_counts())
```

Some movie titles appear multiple times.

For example:

* `Return` → 13 occurrences
* `Beauty and the Beast` → 4 occurrences
* `The Killer` → 4 occurrences
* `The Return` → 4 occurrences
* `Pinocchio` → 4 occurrences

However, the same title can sometimes represent different movies, so **ID should be treated as the primary unique identifier**.

---

## 8. Remove Duplicate Movie Titles

In this cleaning process, duplicate titles are also removed.

```python
df = df.drop_duplicates(subset="Title")

print(df)
```

After removing duplicate titles:

```text
9591 rows × 13 columns
```

---

## 9. Verify Duplicate Titles

```python
print(df["Title"].value_counts())
```

After cleaning, every title appears only once.

This confirms that duplicate movie titles have been removed.

---

## 10. Check Missing Movie IDs

```python
print(df["ID"].isnull().sum())
```

Output:

```text
0
```

There are **no missing movie IDs**.

---

## 11. Check Missing Movie Titles

```python
print(df["Title"].isnull().sum())
```

Output:

```text
0
```

There are **no missing movie titles**.

---

## 12. Check Unknown Movie Titles

Some datasets may contain `"Unknown"` instead of an actual value.

```python
count = (df["Title"].str.lower() == "unknown").sum()

print(count)
```

Output:

```text
1
```

There is **1 movie with an unknown title**.

---

## 13. Check Unknown Directors

```python
count = (df["Director"].str.lower() == "unknown").sum()

print(count)
```

Output:

```text
9022
```

There are **9,022 movies where the director is listed as `"Unknown"`**.

This is a significant amount of missing information and should be considered during analysis.

---

## 14. Save the Cleaned Dataset

After completing the cleaning process, the cleaned dataset is exported as a CSV file.

```python
df.to_csv("movie_cleaned-01_data.csv", index=False)
```

The cleaned dataset is saved as:

```text
movie_cleaned-01_data.csv
```

---

# Data Cleaning Summary

| Cleaning Step                        | Result |
| ------------------------------------ | -----: |
| Original Rows                        | 18,520 |
| Original Columns                     |     13 |
| Rows After Removing Duplicate IDs    |  9,959 |
| Rows After Removing Duplicate Titles |  9,591 |
| Missing IDs                          |      0 |
| Missing Titles                       |      0 |
| Unknown Titles                       |      1 |
| Unknown Directors                    |  9,022 |

## Final Dataset

The final cleaned dataset contains:

**9,591 unique movies and 13 columns.**

The cleaned CSV file is ready for further **Exploratory Data Analysis (EDA), visualization, and Power BI dashboard development.**
````
